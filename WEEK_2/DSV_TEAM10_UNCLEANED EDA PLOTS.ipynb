```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('/content/retail_store_sales (1).csv')
display(df.head())
```

### Initial Exploratory Data Analysis (EDA) Plots

```python
# Plot 1: Distribution of numerical features (histograms)
df.hist(figsize=(15, 10), bins=20)
plt.suptitle('Histograms of Numerical Features', y=1.02)
plt.tight_layout(rect=[0, 0.03, 1, 0.98])
plt.show()
```

```python
# Plot 2: Correlation heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Matrix of Numerical Features')
plt.show()
```

```python
# Plot 3: Box plot of 'Total Spent'
plt.figure(figsize=(10, 6))
sns.boxplot(y=df['Total Spent'])
plt.title('Box Plot of Total Spent')
plt.ylabel('Total Spent')
plt.show()
```

```python
# Plot 11: Daily Sales Trend (Total Spent over time)

df['Transaction Date'] = pd.to_datetime(df['Transaction Date'])
df_daily_sales = df.groupby('Transaction Date')['Total Spent'].sum().reset_index()

plt.figure(figsize=(14, 7))
sns.lineplot(data=df_daily_sales, x='Transaction Date', y='Total Spent')
plt.title('Daily Total Spent Over Time')
plt.xlabel('Date')
plt.ylabel('Total Spent')
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Plot 13: Violin plot of 'Quantity' by 'Category'
plt.figure(figsize=(14, 7))
sns.violinplot(data=df, x='Category', y='Quantity', palette='coolwarm')
plt.title('Quantity Distribution by Category')
plt.xlabel('Category')
plt.ylabel('Quantity')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
# Plot 14: Bar plot of average 'Total Spent' by 'Payment Method'
plt.figure(figsize=(10, 6))
sns.barplot(data=df, x='Payment Method', y='Total Spent', estimator='mean', palette='viridis')
plt.title('Average Total Spent by Payment Method')
plt.xlabel('Payment Method')
plt.ylabel('Average Total Spent')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
# Plot 15: Bar plot of average 'Total Spent' by 'Location'
plt.figure(figsize=(8, 5))
sns.barplot(data=df, x='Location', y='Total Spent', estimator='mean', palette='plasma')
plt.title('Average Total Spent by Location')
plt.xlabel('Location')
plt.ylabel('Average Total Spent')
plt.show()
```

```python
# Plot 16: Count plot of 'Category' with 'Discount Applied' hue
plt.figure(figsize=(14, 7))
sns.countplot(data=df, x='Category', hue='Discount Applied', palette='rocket')
plt.title('Category Distribution by Discount Applied Status')
plt.xlabel('Category')
plt.ylabel('Count')
plt.xticks(rotation=45, ha='right')
plt.legend(title='Discount Applied')
plt.tight_layout()
plt.show()
```

```python
# Plot 17: Histogram of 'Total Spent' with KDE
plt.figure(figsize=(10, 6))
sns.histplot(df['Total Spent'], bins=30, kde=True, color='skyblue')
plt.title('Distribution of Total Spent')
plt.xlabel('Total Spent')
plt.ylabel('Frequency')
plt.tight_layout()
plt.show()
```

```python
# Plot 18: Scatter plot with regression line for 'Price Per Unit' vs 'Total Spent'
plt.figure(figsize=(10, 6))
sns.regplot(data=df, x='Price Per Unit', y='Total Spent', scatter_kws={'alpha':0.3}, line_kws={'color':'red'})
plt.title('Price Per Unit vs Total Spent with Regression Line')
plt.xlabel('Price Per Unit')
plt.ylabel('Total Spent')
plt.show()
```

```python
# Plot 19: Missing values heatmap
plt.figure(figsize=(10, 6))
sns.heatmap(df.isnull(), cbar=False, cmap='viridis')
plt.title('Missing Values Heatmap')
plt.tight_layout()
plt.show()
```

```python
# Plot 20: Bar plot of null values per column
missing_values = df.isnull().sum()
missing_values = missing_values[missing_values > 0]
if not missing_values.empty:
    plt.figure(figsize=(10, 6))
    sns.barplot(x=missing_values.index, y=missing_values.values, palette='magma')
    plt.title('Number of Missing Values Per Column')
    plt.xlabel('Columns')
    plt.ylabel('Number of Missing Values')
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.show()
else:
    print("No missing values found in the DataFrame.")
```

### Final Set of Exploratory Data Analysis (EDA) Plots (21-30)

```python
# Plot 21: Pairplot of numerical features
sns.pairplot(df[['Price Per Unit', 'Quantity', 'Total Spent']], diag_kind='kde')
plt.suptitle('Pair Plot of Numerical Features', y=1.02)
plt.tight_layout(rect=[0, 0.03, 1, 0.98])
plt.show()
```

```python
# Plot 22: Swarm plot of 'Total Spent' by 'Payment Method'
plt.figure(figsize=(12, 7))
sns.swarmplot(data=df, x='Payment Method', y='Total Spent', palette='cubehelix', s=3)
plt.title('Total Spent Distribution by Payment Method (Swarm Plot)')
plt.xlabel('Payment Method')
plt.ylabel('Total Spent')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
# Plot 23: Heatmap of average 'Total Spent' by 'Category' and 'Location'
pivot_table = df.pivot_table(values='Total Spent', index='Category', columns='Location', aggfunc='mean')
plt.figure(figsize=(12, 8))
sns.heatmap(pivot_table, annot=True, cmap='YlGnBu', fmt='.1f')
plt.title('Average Total Spent by Category and Location')
plt.xlabel('Location')
plt.ylabel('Category')
plt.tight_layout()
plt.show()
```

```python
# Plot 24: Monthly Sales Trend
df['Transaction Date'] = pd.to_datetime(df['Transaction Date'])
df['Month'] = df['Transaction Date'].dt.to_period('M')
monthly_sales = df.groupby('Month')['Total Spent'].sum().reset_index()
monthly_sales['Month'] = monthly_sales['Month'].astype(str)

plt.figure(figsize=(14, 7))
sns.lineplot(data=monthly_sales, x='Month', y='Total Spent', marker='o')
plt.title('Monthly Total Spent Over Time')
plt.xlabel('Month')
plt.ylabel('Total Spent')
plt.xticks(rotation=45, ha='right')
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Plot 25: Top 10 Items by Total Sales
top_items = df.groupby('Item')['Total Spent'].sum().nlargest(10).reset_index()
plt.figure(figsize=(12, 7))
sns.barplot(data=top_items, x='Item', y='Total Spent', palette='rocket')
plt.title('Top 10 Items by Total Sales')
plt.xlabel('Item')
plt.ylabel('Total Sales')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
# Plot 26: Pie chart of Payment Method distribution
payment_counts = df['Payment Method'].value_counts()
plt.figure(figsize=(8, 8))
plt.pie(payment_counts, labels=payment_counts.index, autopct='%1.1f%%', startangle=140, colors=sns.color_palette('pastel'))
plt.title('Distribution of Payment Methods')
plt.axis('equal') # Equal aspect ratio ensures that pie is drawn as a circle.
plt.tight_layout()
plt.show()
```

```python
# Plot 27: Box plot of 'Price Per Unit' by 'Location'
plt.figure(figsize=(8, 6))
sns.boxplot(data=df, x='Location', y='Price Per Unit', palette='Set2')
plt.title('Price Per Unit Distribution by Location')
plt.xlabel('Location')
plt.ylabel('Price Per Unit')
plt.show()
```

```python
# Plot 28: Scatter plot of 'Price Per Unit' vs 'Quantity'
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='Price Per Unit', y='Quantity', hue='Category', palette='deep', alpha=0.6)
plt.title('Price Per Unit vs Quantity')
plt.xlabel('Price Per Unit')
plt.ylabel('Quantity')
plt.show()
```

```python
# Plot 29: Categorical scatter plot for 'Discount Applied' vs 'Total Spent'
plt.figure(figsize=(10, 6))
sns.catplot(data=df, x='Discount Applied', y='Total Spent', kind='swarm', hue='Location', palette='viridis', height=6, aspect=1.5)
plt.title('Total Spent with and without Discount, by Location')
plt.xlabel('Discount Applied')
plt.ylabel('Total Spent')
plt.tight_layout()
plt.show()
```

```python
# Plot 30: KDE plot for 'Price Per Unit'
plt.figure(figsize=(10, 6))
sns.kdeplot(df['Price Per Unit'], fill=True, color='purple')
plt.title('Kernel Density Estimate of Price Per Unit')
plt.xlabel('Price Per Unit')
plt.ylabel('Density')
plt.tight_layout()
plt.show()
```
