```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Load the dataset
df = pd.read_csv('/content/CLEANED_DATASET (1).csv')

# Clean column names by removing leading/trailing spaces and replacing spaces with underscores
df.columns = df.columns.str.strip().str.replace(' ', '_')

# Display the first few rows and information about the dataframe
print("DataFrame Info:")
df.info()
print("\nFirst 5 rows of the dataset:")
display(df.head())
```

## EDA Plots

```python
# Plot 1: Distribution of 'Total_Spent'
plt.figure(figsize=(10, 6))
sns.histplot(df['Total_Spent'], kde=True)
plt.title('Distribution of Total Spent')
plt.xlabel('Total Spent')
plt.ylabel('Frequency')
plt.show()
```

```python
# Plot 2: Distribution of 'Price_Per_Unit'
plt.figure(figsize=(10, 6))
sns.histplot(df['Price_Per_Unit'], kde=True)
plt.title('Distribution of Price Per Unit')
plt.xlabel('Price Per Unit')
plt.ylabel('Frequency')
plt.show()
```

```python
# Plot 3: Count of products by 'Category'
plt.figure(figsize=(12, 7))
sns.countplot(y='Category', data=df, order=df['Category'].value_counts().index)
plt.title('Count of Products by Category')
plt.xlabel('Count')
plt.ylabel('Category')
plt.show()
```

```python
# Plot 4: Average 'Total_Spent' by 'Category'
plt.figure(figsize=(12, 7))
sns.barplot(x='Total_Spent', y='Category', data=df, estimator=lambda x: x.mean(), ci=None, order=df.groupby('Category')['Total_Spent'].mean().sort_values(ascending=False).index)
plt.title('Average Total Spent by Category')
plt.xlabel('Average Total Spent')
plt.ylabel('Category')
plt.show()
```

```python
# Plot 5: Relationship between 'Price_Per_Unit' and 'Quantity' using a scatter plot
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Quantity', y='Price_Per_Unit', data=df)
plt.title('Price Per Unit vs. Quantity')
plt.xlabel('Quantity')
plt.ylabel('Price Per Unit')
plt.show()
```

```python
# Plot 6: Box plot of 'Total_Spent' by 'Category' to show price distribution
plt.figure(figsize=(14, 8))
sns.boxplot(x='Total_Spent', y='Category', data=df)
plt.title('Total Spent Distribution by Category')
plt.xlabel('Total Spent')
plt.ylabel('Category')
plt.show()
```

```python
# Plot 7: Count of products by 'Payment_Method'
plt.figure(figsize=(10, 6))
sns.countplot(y='Payment_Method', data=df, order=df['Payment_Method'].value_counts().index)
plt.title('Count of Products by Payment Method')
plt.xlabel('Count')
plt.ylabel('Payment Method')
plt.show()
```

```python
# Plot 8: Count of products by 'Location'
plt.figure(figsize=(10, 6))
sns.countplot(y='Location', data=df, order=df['Location'].value_counts().index)
plt.title('Count of Products by Location')
plt.xlabel('Count')
plt.ylabel('Location')
plt.show()
```

```python
# Plot 9: Count of 'Discount_Applied' products
plt.figure(figsize=(8, 5))
sns.countplot(x='Discount_Applied', data=df)
plt.title('Count of Products with Discount Applied')
plt.xlabel('Discount Applied')
plt.ylabel('Count')
plt.show()
```

```python
# Plot 10: Box plot of 'Total_Spent' for 'Discount_Applied' vs 'Not Discount_Applied'
plt.figure(figsize=(8, 6))
sns.boxplot(x='Discount_Applied', y='Total_Spent', data=df)
plt.title('Total Spent Distribution for Discount Applied vs. Not Discount Applied Products')
plt.xlabel('Discount Applied')
plt.ylabel('Total Spent')
plt.show()
```
